```python
import os
import h5py
import numpy as np
import matplotlib.pyplot as plt
from scipy.interpolate import interp1d
import matplotlib.cm as cm
from scipy.stats import mannwhitneyu

summary_swimming_durations = []

def load_fish_data(path_step, fish_id):
    all_runs = {
        'binary': [],
        't_c': [],
        'vfish_c': [],
        'mean_thrust_c': [],
        'thrust_thr': [],
        'mask_movetail': [],
        'tvals': [],  # optional
    }

    for file in os.listdir(path_step):
        if file.startswith(fish_id + '_'):
            file_path = os.path.join(path_step, file)
            if not os.path.isfile(file_path):
                continue
            try:
                with h5py.File(file_path, "r") as hdf5_file:
                    binary = hdf5_file['post-processing/binary_output'][:]
                    t_c = hdf5_file['post-processing/t_c'][:]
                    vfish_c = hdf5_file['post-processing/speed_smooth'][:]
                    mean_thrust_c = hdf5_file['post-processing/mean_thrust_c'][:]
                    thrust_thr = hdf5_file['post-processing/thrust_thr'][()]
                    tvals = hdf5_file['post-processing/tvals'][:]
            except Exception as e:
                print(f"⚠️ Could not read {file}: {e}")
                continue

            all_runs['binary'].append(binary)
            all_runs['t_c'].append(t_c)
            all_runs['vfish_c'].append(vfish_c)
            all_runs['mean_thrust_c'].append(mean_thrust_c)
            all_runs['thrust_thr'].append(thrust_thr)
            all_runs['mask_movetail'].append(binary.astype(bool))
            all_runs['tvals'].append(tvals)

    return all_runs

def compute_all_tau(mask_list):
    tau_on_all, tau_off_all = [], []
    for mask in mask_list:
        if mask is not None and np.any(mask):  # Skip empty or all-False masks
            tau_on, tau_off = compute_tau_on_off(mask)
            tau_on_all.append(tau_on)
            tau_off_all.append(tau_off)
    if len(tau_on_all) == 0:
        return np.array([]), np.array([])
    return np.concatenate(tau_on_all), np.concatenate(tau_off_all)

# List all fish_ids by stripping the suffix from filename
def extract_fish_ids(path):
    fish_ids = set()
    for file in os.listdir(path):
        if file.startswith('zf_') and not file.startswith('.'):
            parts = file.split('_')
            if parts[-1].isdigit():
                fish_id = '_'.join(parts[:-1])  # drop ONLY the last part
                fish_ids.add(fish_id)
    return sorted(fish_ids)

# Paths
root_path = str(DATA_VISCOSITY_ZF)
path_control = os.path.join(root_path, 'pre')
path_viscosity = os.path.join(root_path, 'dextran')

# Extract all fish IDs present in both folders
fish_ids = list(set(extract_fish_ids(path_control)) & set(extract_fish_ids(path_viscosity)))
print(f"Found {len(fish_ids)} fish:", fish_ids)

# Loop over all fish
for i, fish_id in enumerate(fish_ids):
    print(f"\nProcessing {fish_id} ({i+1}/{len(fish_ids)})")
    print(fish_id)
    # Load data
    control_data = load_fish_data(path_control, fish_id)
    viscosity_data = load_fish_data(path_viscosity, fish_id)
    
    # Compute τ_on for control and viscosity
    tau_on_control, _ = compute_all_tau(control_data['mask_movetail'])
    tau_on_viscosity, _ = compute_all_tau(viscosity_data['mask_movetail'])
    # Compute statistical test between the two distributions
    try:
        stat, pval = mannwhitneyu(tau_on_control, tau_on_viscosity, alternative='two-sided')
    except ValueError:
        stat, pval = np.nan, np.nan

    # Skip fish if either array is empty
    if tau_on_control.size == 0 or tau_on_viscosity.size == 0:
        print(f"⚠️ Skipping {fish_id} — empty τ_on array(s)")
        continue

    # Compute stats
    median_control = np.median(tau_on_control)
    q1_control, q3_control = np.percentile(tau_on_control, [25, 75])

    median_visc = np.median(tau_on_viscosity)
    q1_visc, q3_visc = np.percentile(tau_on_viscosity, [25, 75])
    
    mean_control = np.mean(tau_on_control)
    std_control = np.std(tau_on_control)

    mean_viscosity = np.mean(tau_on_viscosity)
    std_viscosity = np.std(tau_on_viscosity)
        
    # Store summary
    summary_swimming_durations.append({
        'fish_id': fish_id,
        'median_control': median_control,
        'iqr_control': (q1_control, q3_control),
        'mean_control': mean_control,
        'std_control': std_control,
        'tau_on_control': tau_on_control,
        'median_viscosity': median_visc,
        'iqr_viscosity': (q1_visc, q3_visc),
        'mean_viscosity': mean_viscosity,
        'std_viscosity': std_viscosity,
        'tau_on_viscosity': tau_on_viscosity,
        'p_value': pval
        
    })
```

```python
def compute_all_vmax_onsets(mask_list, vfish_list, dt=1/300, window_ms=400):
    vmax_all = []
    window_samples = int(window_ms / 1000 / dt)

    for mask, vfish in zip(mask_list, vfish_list):
        if mask is None or vfish is None:
            continue
        if not np.any(mask):
            continue

        mask = np.array(mask, dtype=bool)
        vfish = np.array(vfish)
        onsets = (mask[1:] & ~mask[:-1]).nonzero()[0] + 1

        vmax_list = []
        for onset in onsets:
            end = onset + window_samples
            if end <= len(vfish):
                segment = vfish[onset:end]
                vmax = np.nanmax(segment)
                vmax_list.append(vmax)
        vmax_all.extend(vmax_list)

    return np.array(vmax_all)
```

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.signal import find_peaks

def compute_all_vpeaks(mask_list, vfish_list, dt=1/300,
                       height=None, distance=None, prominence=None):
    """
    Compute peaks of vfish within mask==1 periods for a list of runs.

    Parameters
    ----------
    mask_list : list of 1D arrays
        Binary masks (0/1 or bool), one per run.
    vfish_list : list of 1D arrays
        Velocity traces, same length as corresponding masks.
    dt : float
        Time step between samples (s).
    height, distance, prominence :
        Passed directly to scipy.signal.find_peaks.

    Returns
    -------
    all_peaks : np.ndarray
        Concatenated array of peak values across all runs.
    per_run : list of dict
        Per-run info for plotting:
            {
                'run_idx': int,
                't': 1D array of times,
                'v': 1D array of velocities,
                'mask': 1D bool array,
                'peaks_idx': 1D array of indices in original signal,
                'peaks_t': 1D array of times at peaks,
                'peaks_v': 1D array of values at peaks
            }
        For runs with no data or no peaks, peaks_* arrays are empty.
    """
    all_peaks = []
    per_run = []

    for run_idx, (mask, vfish) in enumerate(zip(mask_list, vfish_list)):
        if mask is None or vfish is None:
            per_run.append({
                'run_idx': run_idx,
                't': None,
                'v': None,
                'mask': None,
                'peaks_idx': np.array([], dtype=int),
                'peaks_t': np.array([], dtype=float),
                'peaks_v': np.array([], dtype=float),
            })
            continue

        mask = np.asarray(mask, dtype=bool)
        vfish = np.asarray(vfish, dtype=float)

        if len(mask) != len(vfish):
            raise ValueError(f"mask and vfish must have same length, got {len(mask)} and {len(vfish)}")

        if not np.any(mask):
            # No tail movement
            t = np.arange(len(vfish)) * dt
            per_run.append({
                'run_idx': run_idx,
                't': t,
                'v': vfish,
                'mask': mask,
                'peaks_idx': np.array([], dtype=int),
                'peaks_t': np.array([], dtype=float),
                'peaks_v': np.array([], dtype=float),
            })
            continue

        t = np.arange(len(vfish)) * dt

        # Masked data (only where tail moves)
        t_masked = t[mask]
        v_masked = vfish[mask]

        # Find peaks on masked signal
        peaks_idx_masked, _ = find_peaks(
            v_masked,
            height=height,
            distance=distance,
            prominence=prominence
        )

        if peaks_idx_masked.size == 0:
            per_run.append({
                'run_idx': run_idx,
                't': t,
                'v': vfish,
                'mask': mask,
                'peaks_idx': np.array([], dtype=int),
                'peaks_t': np.array([], dtype=float),
                'peaks_v': np.array([], dtype=float),
            })
            continue

        # Map back to original indices
        orig_idx = np.where(mask)[0][peaks_idx_masked]
        peaks_t = t[orig_idx]
        peaks_v = vfish[orig_idx]

        all_peaks.extend(peaks_v)

        per_run.append({
            'run_idx': run_idx,
            't': t,
            'v': vfish,
            'mask': mask,
            'peaks_idx': orig_idx,
            'peaks_t': peaks_t,
            'peaks_v': peaks_v,
        })

    return np.asarray(all_peaks), per_run
```

```python
def plot_peaks_for_fish_condition(fish_id, per_run_info, condition_name,
                                  figsize=(8, 3)):
    """
    For a given fish and condition (control/viscosity),
    make one figure per run: raw vfish + peaks marked.
    """
    for info in per_run_info:
        t = info['t']
        v = info['v']
        if t is None or v is None:
            continue  # run was None

        peaks_t = info['peaks_t']
        peaks_v = info['peaks_v']
        run_idx = info['run_idx']

        plt.figure(figsize=figsize)
        plt.plot(t, v, label='vfish (raw)', alpha=0.6)
        plt.scatter(peaks_t, peaks_v, s=20, label='peaks')

        # Optional: highlight masked region lightly
        if info['mask'] is not None and np.any(info['mask']):
            # where mask==1, fill under curve
            plt.plot(t[info['mask']], v[info['mask']],
                     linewidth=1.5, label='mask==1')

        plt.xlabel("Time (s)")
        plt.ylabel("v_fish")
        plt.title(f"{fish_id} – {condition_name} – run {run_idx}")
        plt.legend()
        plt.grid(alpha=0.2)
        plt.tight_layout()
```
